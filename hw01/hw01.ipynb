````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?

````python
# База
````

````python
#1.
cities_clean = []
for city in cities:
  cities_clean.append(city.strip().capitalize())
print(cities_clean)
````

````python
#2.
amounts_int = [int(amount) for amount in amounts]
print(amounts_int)
````

````python
#3.
total_orders = len(amounts_int)
total_sum = sum(amounts_int)
average_check = round(total_sum / total_orders)
print(f"Общее число заказов: {total_orders}")
print(f"Общая сумма: {total_sum}")
print(f"Средний чек: {average_check}")
````

````python
#4.
not_paid_orders = [amount for amount, paid in zip(amounts_int, paid) if not paid]
total_not_paid = sum(not_paid_orders)
print(f"Не оплачено заказов: {len(not_paid_orders)}")
print(f"Общая сумма не оплаченных заказов: {total_not_paid}")
````

````python
#5.
for city, amount, p in zip(cities_clean, amounts_int, paid):
  status = "оплачено" if p else "не оплачено"
  print(f"{city} | {amount} | {status}")
````

## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
ala_rev = 0
ast_rev = 0

for c, p, s in zip(cities_clean, amounts_int, paid):
  if c == "Almaty" and s:
    ala_rev += p
  elif c == "Astana" and s:
    ast_rev += p

print(f"Выручка по Алматы: {ala_rev}")
print(f"Выручка по Астане: {ast_rev}")
````

````python
maxi_amount = max(amounts_int)
maxi_index = amounts_int.index(maxi_amount)
print(f"Номер самого крупного оплаченного заказа: {maxi_index}")
print(f"Город самого крупного оплаченного заказа: {cities_clean[maxi_index]}")
````

````python
sat = "сб"
sun = "вс"
sat_count = 0
sun_count = 0

for day in days:
  if day == sat:
    sat_count += 1
  elif day == sun:
    sun_count += 1

total_days = len(days)
sat_percent = (sat_count / total_days) * 100
sun_percent = (sun_count / total_days) * 100

print(f"Доля заказов выходного дня (сб): {sat_percent:.2f}%")
print(f"Доля заказов выходного дня (вс): {sun_percent:.2f}%")
````

````python
bank = 0
payment = 15000
month = 0

while bank < 400000:
  bank += payment
  month += 1
  payment += 1000

print(f"Студент накопит через {month} месяцев")
print(f"Итого в копилке: {bank}")
````

````python
# Челлендж
````

````python
unique_cities = []
revenue_by_city = []

for city, amount, is_paid in zip(cities_clean, amounts_int, paid):
  if is_paid:
    if city not in unique_cities:
      unique_cities.append(city)
      revenue_by_city.append(amount)
    else:
      index = unique_cities.index(city)
      revenue_by_city[index] += amount

n = len(revenue_by_city)
for i in range(n):
  for j in range(0, n - i - 1):
    if revenue_by_city[j] < revenue_by_city[j + 1]:
      revenue_by_city[j], revenue_by_city[j + 1] = revenue_by_city[j + 1], revenue_by_city[j]
      unique_cities[j], unique_cities[j + 1] = unique_cities[j + 1], unique_cities[j]


for city, rev in zip(unique_cities, revenue_by_city):
  print(f"{city}: {rev}")
````

````python
if len(revenue_by_city) % 2 == 1:
  mid = len(revenue_by_city) // 2
  median = revenue_by_city[mid]
else:
  mid1 = len(revenue_by_city) // 2
  mid2 = mid1 +1
  median = (revenue_by_city[mid1] + revenue_by_city[mid2]) / 2

print(f"Медиана суммы заказа: {median}")
````
